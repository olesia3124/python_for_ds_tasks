Для виведення результатів фільтрації:
- зберегти результат у окрему змінну

```python
# Фільтрація всіх записів з рахунками більше 20 доларів (total_bill > 20) набору даних Tips

# Збереження всіх записів з рахунками більше 20 доларів набору даних Tips в окрему змінну
total_bill_above_20 = df[df['total_bill'] > 20]

# Показ кількості записів у відфільтрованому наборі
print(f"Кількість записів з рахунками більше 20 доларів: {len(total_bill_above_20)}")

# Виведення перших 5 записів відфільтрованого набору
total_bill_above_20.head()
```

```python
# Фільтрація всіх записів з відвідувачами-жінками (sex == 'Female') набору даних Tips

# Збереження всіх записів з відвідувачами-жінками набору даних Tips в окрему змінну
sex_female = df[df['sex'] == 'Female']

# Показ кількості записів у відфільтрованому наборі
print(f"Кількість записів з відвідувачами-жінками: {len(sex_female)}")

# Виведення перших 5 записів відфільтрованого набору
sex_female.head()
```

```python
# Фільтрація всіх записів з рахунками на вечерю (time == 'Dinner') набору даних Tips

# Збереження всіх записів з рахунками на вечерю набору даних Tips в окрему змінну
time_dinner = df[df['time'] == 'Dinner']

# Показ кількості записів у відфільтрованому наборі
print(f"Кількість записів з рахунками на вечерю: {len(time_dinner)}")

# Виведення перших 5 записів відфільтрованого набору
time_dinner.head()
```

```python
# Фільтрація всіх записів з рахунками на вечерю більше 30 доларів набору даних Tips

# Збереження всіх записів з рахунками на вечерю більше 30 доларів набору даних Tips в окрему змінну
time_dinner_and_total_bill_above_30 = df[(df['time'] == 'Dinner') & (df['total_bill'] > 30)]

# Показ кількості записів у відфільтрованому наборі
print(f"Кількість записів з рахунками на вечерю більше 30 доларів: {len(time_dinner_and_total_bill_above_30)}")

# Виведення перших 5 записів відфільтрованого набору
time_dinner_and_total_bill_above_30.head()
```

Обчисліть (кожен пункт оцінюється в 0.5 бала):

```python
# Обчислення кількості пропущених значень у всіх колонках на наборі даних Tips
null_counted = df.isnull().sum()
print("Кількість пропущених значень у всіх колонках:")
null_counted
```

```python
# Обчислення кількості рахунків за обіди і за вечері на наборі даних Tips

# Фільтрація всіх записів з рахунками за обіди і за вечері набору даних Tips
lunches_and_dinners = df[df['time'].isin(['Lunch', 'Dinner'])]
display(lunches_and_dinners)

# Обчислення кількості рахунків у відфільтрованому наборі
print(f"Кількість записів з рахунками за обіди і за вечері: {len(lunches_and_dinners)}")
```

**5. Статистика із групуванням**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній рахунок за статтю (groupby `sex`)
- Середні чайові за днем тижня (groupby `day`)
- Скільки відвідувачів у кожен день

```python
# Обчислення середнього рахунку за статтю (groupby sex) на наборі даних Tips
mean_total_bill_by_class = df.groupby('sex')[['total_bill']].mean()
print("Середній рахунок за статтю:")
mean_total_bill_by_class
```

```python
# Обчислення середніх чайових за днем тижня (groupby day) на наборі даних Tips
mean_tip_by_day = df.groupby('day')[['tip']].mean()
print("Середні чайові за днем тижня:")
mean_tip_by_day
```

```python
# Обчислення кількості відвідувачів у кожен день тижня на наборі даних Tips
sum_size_by_day = df.groupby('day')[['size']].sum()
print("Кількість відвідувачів за днем тижня:")
sum_size_by_day
```

**7. Сортування**

Відсортуйте датасет і виведіть результат сортування в кожному випадку окремо (кожен пункт оцінюється в 0.5 бала):
- За розміром рахунку (зростання)

Додатково виведіть топ-5 найбільших чайових використовуючи метод `nlargest`.

```python
# Сортування набору даних Tips за розміром рахунку (за зростанням)
df_sorted_total_bill = df.sort_values('total_bill')
df_sorted_total_bill
```

```python
# Сортування набору даних Tips за розміром чайових (за спаданням)
df_sorted_tip_desc = df.sort_values('tip', ascending = False)
df_sorted_tip_desc
```

```python
# Виведення топ-5 найбільших чайових, використовуючи метод nlargest
largest_tips = df.nlargest(5, 'tip')
largest_tips
```
