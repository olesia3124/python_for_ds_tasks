**1. Цикл while**

Задано змінну `counter`, яка рівна -3.

Напишіть програму, використовуючи `while`, яка в циклі додає до змінної `counter` +1 і зупиняється, коли змінна `counter` стане суворо більше за 5. На кожній ітерації циклу виводьте значення змінної `counter`.



```python
# Програма для виведення лічильника для значень <= 5
counter = -3
while counter <= 5:
    print(f"Лічильник: {counter}")
    counter +=1
```

**2. List comprehension - прості операції**

Дано список чисел:

```python
numbers = [11, 4, 9, 16, 7, 5]
```


Створіть нові списки використовуючи list comprehension, які містять (кожен пункт оцінюється в 0.5 бала):
- Квадрати всіх чисел
- Тільки парні числа
- Всі числа помножені на 10

Виведіть новоутворені списки.

```python
# Програма для обчислення квадратів чисел зі списку numbers

# Створення списку квадратів чисел зі списку numbers
square_numbers = [x**2 for x in numbers]

# Виведення списку квадратів чисел зі списку numbers
print(f"Квадрати чисел зі списку numbers: {square_numbers}")
```

```python
# Програма для виведення парних чисел зі списку numbers

# Створення списку парних чисел зі списку numbers
even_numbers = [x for x in numbers if x % 2 == 0]

# Виведення списку парних чисел зі списку numbers
print(f"Парні числа зі списку numbers: {even_numbers}")
```

```python
# Програма для множення на 10 чисел зі списку numbers

# Створення списку чисел зі списку numbers, помножених на 10
x10_numbers = [x*10 for x in numbers]

# Виведення списку чисел зі списку numbers, помножених на 10
print(f"Числа зі списку numbers, помножені на 10: {x10_numbers}")
```

**3. List comprehension - комбінована логіка**

Дано список чисел від 1 до 10. Створіть новий список, де виконується наступна логіка:
- Якщо число парне - помножте його на 2
- Якщо непарне - віднімніть 1

Виведіть результуючий список на екран.

```python
numbers = list(range(1, 11))
```

```python
# Програма для виконання операцій над елементами списку numbers залежно від умови

# Створення списку, де парні елементи numbers домножені на 2, а непарні - зменшені на 1
numbers_after_if = [x*2 if x % 2 == 0 else x - 1 for x in numbers]

# Виведення списку, де парні елементи numbers домножені на 2, а непарні - зменшені на 1
print(f"Числа зі списку numbers, помножені на 2 (для парних) та зменшені на 1 (для непарних): {numbers_after_if}")
```
