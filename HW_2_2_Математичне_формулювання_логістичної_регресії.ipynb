#### Завдання 1: Реалізація функції сигмоїди
1. З використанням `numpy` напишіть функцію `sigmoid(z)` для обчислення значення сигмоїди згідно з формулою:
   $$
   \sigma(z) = \frac{1}{1 + e^{-z}}
   $$
2. Використовуючи цю функцію, обчисліть значення сигмоїди для наступних даних: $ z = [-2, -1, 0, 1, 2] $. Виведіть результат обчислень.


```python
# Імпортування потрібних бібліотек
import numpy as np
```

```python
# Задання функції сигмоїди з використанням numpy

# Створення функції сигмоїди
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Обчислення значень сигмоїди для вектора аргументів z
z = np.array([-2, -1, 0, 1, 2])
sigmoid_results = sigmoid(z)

# Виведення результату обчислень
print(f"Значення функції сигмоїди для вектора z = [-2, -1, 0, 1, 2]:\n{sigmoid_results}")
```



#### Завдання 2: Реалізація функції гіпотези для логістичної регресії
1. Напишіть функцію `hypothesis(theta, X)`, яка обчислює гіпотезу для логістичної регресії, використовуючи функцію сигмоїди. Формула гіпотези:
   $$
   h_\theta(x) = \sigma(\theta^T x) = \frac{1}{1 + e^{-\theta^T x}}
   $$
2. Використайте функцію `hypothesis` для обчислення значень гіпотези для наступних даних:
   
   $\theta = [0.5, -0.5]$
   
   $X = \begin{bmatrix} 1 & 2 \\ 1 & -1 \\ 1 & 0 \\ 1 & 1 \end{bmatrix}$

  Виведіть результат обчислень.


```python
# Задання функції гіпотези для логістичної регресії з використанням функції сигмоїди

# Створення функції гіпотези
def hypothesis(theta, X):
    return sigmoid(np.dot(X, theta.T))

# Обчислення значень функції гіпотези для заданих аргументів 
theta = np.array([0.5, -0.5])
X = np.array([[1, 2],
              [1, -1],
              [1, 0],
              [1, 1]])
hypothesis_results = hypothesis(theta, X)

# Виведення результату обчислень
print(f"Значення функції гіпотези логістичної регресії для заданих аргументів theta та X:\n{hypothesis_results}")
```

#### Завдання 3: Реалізація функції для підрахунку градієнтів функції втрат
1. Напишіть функцію `compute_gradient(theta, X, y)`, яка обчислює градієнти функції втрат для логістичної регресії. Формула для обчислення градієнта:
   $$
   \frac{\partial L(\theta)}{\partial \theta_j} = \frac{1}{m} \sum_{i=1}^{m} \left[ (h_\theta(x^{(i)}) - y^{(i)}) x_j^{(i)} \right]
   $$
2. Використайте функцію `compute_gradient` для обчислення градієнтів для наступних даних:

  $\theta = [0.5, -0.5]$

  $X = \begin{bmatrix} 1 & 2 \\ 1 & -1 \\ 1 & 0 \\ 1 & 1 \end{bmatrix}$

  $y = [1, 0, 1, 0]$

  Виведіть результат обчислень.

```python
# Задання функції для підрахунку градієнтів функції втрат з використанням numpy

# Створення функції для підрахунку градієнтів функції втрат
def compute_gradient(theta, X, y):
    """
    Обчислює градієнти функції втрат для логістичної регресії, де:
    X: матриця ознак розмірності (m, n);
    y: вектор істинних значень розмірності (m,);
    theta: вектор параметрів розмірності (n,).
    """
    m = X.shape[0]  # Кількість навчальних прикладів
    
    # Обчислення передбачень гіпотези hypothesis(theta, x) для всіх параметрів
    h = hypothesis(theta, X)
    
    # Обчислення градієнтів для всіх параметрів одночасно
    gradient = (1 / m) * np.dot(X.T, (h - y))
    return gradient

# Обчислення значень функції для підрахунку градієнтів для заданих аргументів 
theta = np.array([0.5, -0.5])
X = np.array([[1, 2],
              [1, -1],
              [1, 0],
              [1, 1]])
y = np.array([1, 0, 1, 0])
gradient_results = compute_gradient(theta, X, y)

# Виведення результату обчислень
print(f"Значення функції для підрахунку градієнтів функції втрат для заданих аргументів theta, X та y:\n{gradient_results}")
```


#### Завдання 4: Реалізація повного батч градієнтного спуску

**Задача:**
1. Напишіть функцію `full_batch_gradient_descent(X, y, lr=0.1, epochs=100)`, яка реалізує алгоритм Full градієнтного спуску для логістичної регресії. Використовуйте такі формули:
   - Гіпотеза: $ h_\theta(x) = \sigma(\theta^T x) $
   - Оновлення параметрів: $ \theta_j := \theta_j - \alpha \frac{\partial L(\theta)}{\partial \theta_j} $
2. Використайте функцію `full_batch_gradient_descent` для обчислення параметрів моделі на наступних даних:

  $X = \begin{bmatrix} 1 & 2 \\ 1 & -1 \\ 1 & 0 \\ 1 & 1 \end{bmatrix}$

  $y = [1, 0, 1, 0]$

  Увага! Матриця $X$ вже має стовпець одиниць і передбачається, що це. - стовпець для intercept - параметра зсуву.

  Виведіть результат обчислень.


```python
# 4.1. Задання функції, яка реалізує алгоритм повного градієнтного спуску для логістичної регресії

# Створення функції повного градієнтного спуску для логістичної регресії
def full_batch_gradient_descent(X, y, lr = 0.1, epochs = 100):
    """
    Реалізує повний градієнтний спуск для логістичної регресії, де:
    
    X: матриця ознак (з урахуванням стовпця одиниць) розмірності (m, n);
    y: вектор істинних міток розмірності (m,);
    lr: швидкість навчання (learning rate);
    epochs: кількість ітерацій (епох) навчання.
    
    Повертає:
    theta: оптимальні параметри моделі розмірності (n,).
    """
    # Ініціалізація ваг theta нулями (розмірність відповідає кількості ознак n)
    n = X.shape[1]
    theta = np.zeros(n)
    
    # Цикл навчання по епохах
    for epoch in range(epochs):
        # Обчислення градієнту для поточної theta на всьому наборі даних (Full Batch)
        gradient = compute_gradient(theta, X, y)
        
        # Оновлення параметрів theta у протилежному до градієнта напрямку
        theta = theta - lr * gradient    
    return theta
```

```python
# 4.2. Використання функції full_batch_gradient_descent() для обчислення параметрів моделі на заданих даних

# Задані аргументи
X = np.array([[1, 2],
              [1, -1],
              [1, 0],
              [1, 1]])
y = np.array([1, 0, 1, 0])

# Запуск градієнтного спуску
optimal_theta = full_batch_gradient_descent(X, y)

# Виведення результату обчислень
print(f"Оптимальний вектор параметрів theta: {optimal_theta}")
```

#### Завдання 5. Обчислення точності моделі

1. Напишіть функцію `predict_proba(theta, X)`, яка використовує знайдені параметри $\theta$ для обчислення ймовірностей належності поточного прикладу з даних до класу $y=1$ на основі значень $\sigma(\theta^T x)$.

2. Напишіть функцію `predict(theta, X, threshold=0.5)`, яка обчислює клас з передбаченої імовірності належності екземпляра до класу 1 з порогом 0.5. Тобто якщо ймовірність менше 0.5, то передбачаємо клас 0, інакше клас 1.

3. Напишіть функцію `accuracy(y_true, y_pred)`, яка обчислює точність моделі, визначивши частку правильно передбачених класів.

  Формула метрики Accuracy:
  $$
  \text{Accuracy} = \frac{\sum_{i=1}^{m} I(\hat{{y}^{(i)}} = y^{(i)})}{m}
  $$

  де $\hat{{y}^{(i)}}$ - передбачене значення класу, $I$ - індикаторна функція (яка дорівнює 1, якщо умова виконується, і 0 - якщо ні), $m$ - кількість прикладів.

4. Обчисліть з використанням даних в завданні 4 $X$, $y$ та обчислених коефіцієнтах $\theta$ та виведіть на екран:
  - передбачені моделлю імовірності належності кожного з екземплярів в матриці `X` до класу 1
  - класи кожного екземпляра з матриці `X`
  - точність моделі.

```python
# 5.1. Задання функції, яка використовує знайдені параметри theta для обчислення ймовірностей належності поточного прикладу 
# з даних до класу y = 1 на основі значень сигмоїди

# Створення функції predict_proba()
def predict_proba(theta, X):
    """
    Обчислює ймовірність належності кожного прикладу до класу 1, де:
    X: матриця ознак розмірності (m, n);
    theta: вектор знайдених параметрів розмірності (n,);
    Повертає: вектор ймовірностей розмірності (m,).
    """
    # Множення X на theta (скалярний добуток) та взяття сигмоїди
    return sigmoid(np.dot(X, theta))
```

```python
# 5.2. Задання функції, яка обчислює клас з передбаченої імовірності належності екземпляра до класу 1 з порогом 0.5

# Створення функції predict()
def predict(theta, X, threshold = 0.5):
    """
    Прогнозує конкретні класи (0 або 1) для кожного прикладу, де:
    theta: вектор знайдених параметрів розмірності (n,);
    X: матриця ознак розмірності (m, n);
    threshold: поріг відсікання ймовірності (за замовчуванням 0.5).
    
    Повертає: вектор бінарних прогнозів (0 або 1) розмірності (m,).
    """
    # Отримання ймовірностей за допомогою раніше створеної функції
    probabilities = predict_proba(theta, X)
    
    # Якщо ймовірність >= threshold, то True (1), інакше False (0)
    return (probabilities >= threshold).astype(int) # .astype(int) автоматично перетворює True в 1, а False в 0
```

```python
# 5.3. Задання функції, яка обчислює точність моделі, визначивши частку правильно передбачених класів

# Створення функції accuracy(y_true, y_pred)
def accuracy(y_true, y_pred):
    """
    Обчислює точність моделі (Accuracy), де:
    y_true: вектор істинних значень класів (0 або 1) розмірності (m,);
    y_pred: вектор передбачених моделлю класів (0 або 1) розмірності (m,).
    
    Повертає: значення точності від 0.0 до 1.0.
    """
    # Обчислення m (загальної кількості прикладів)
    m = len(y_true)
    
    # Обчислення кількості правильних передбачень (де y_true == y_pred) та ділення на загальну кількість m
    return np.sum(y_true == y_pred) / m
```

```python
# 5.4. Обчислення з використанням даних в завданні 4 X, y й обчислених коефіцієнтів theta та виведення на екран:

# -- передбачених моделлю ймовірностей належності кожного з екземплярів в матриці X до класу 1

# Задані вхідні дані
X = np.array([[1, 2],
              [1, -1],
              [1, 0],
              [1, 1]])
y = np.array([1, 0, 1, 0])

# Знайдені градієнтним спуском у пункті 4 оптимальні параметри theta
theta = optimal_theta

# Обчислення ймовірностей
probabilities = predict_proba(theta, X)

# Виведення результату
print("Обчислені ймовірності для кожного прикладу:")
for i, prob in enumerate(probabilities):
    print(f"Приклад № {i+1}: {X[i]} -> Ймовірність класу 1: {prob:.4f} (Істинний клас y: {y[i]})")


# -- класів кожного екземпляра з матриці X

# Отримання фінальних бінарних класів
predictions = predict(theta, X, threshold = 0.5)

# Виведення результату
print(f"Істинні класи (y):  {y}")
print(f"Прогнозовані класи: {predictions}")


# -- точності моделі

# Дані з попереднього кроку
y_true = y
y_pred = predictions

# Обчислення метрики
acc_score = accuracy(y_true, y_pred)

# Виведення результату
print(f"Точність моделі (Accuracy): {acc_score} ({acc_score * 100:.1f}%)")
```
