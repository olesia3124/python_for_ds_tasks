**Про датасет:**

Tips - це датасет, який містить інформацію про рахунки та чайові в ресторані.

**Колонки датасету:**
- `total_bill` - загальна сума рахунку (долари)
- `tip` - розмір чайових (долари)
- `sex` - стать відвідувача (Male/Female)
- `smoker` - чи курить (Yes/No)
- `day` - день тижня (Thur/Fri/Sat/Sun)
- `time` - час відвідування (Lunch/Dinner)
- `size` - розмір групи (кількість людей)

**Джерело:** Оригінальні дані зібрані в одному з американських ресторанів у 1990-х роках. Скачати і переглянути набір даних можна тут: https://www.kaggle.com/datasets/sakshisatre/tips-dataset

**1. Гістограма розподілу рахунків**
Завантажте датасет Tips (Чайові) з файла у матеріалах `tips.csv`.
Створіть гістограму для колонки `total_bill` з 20 bins.
Опціонально можна додати підписи осей ОХ, ОУ, а також назву графіку.

```python
# Імпорт бібліотеки pandas під скороченим псевдонімом pd
import pandas as pd

# Імпорт бібліотеки matplotlib під скороченим псевдонімом plt
import matplotlib.pyplot as plt

# Завантаження набору даних Tips з файлу csv
df = pd.read_csv('tips.csv')
```

```python
# Створення гістограми з 20 стовпчиками для колонки total_bill набору даних Tips
df['total_bill'].hist(bins = 20)

# Створення назви графіка
plt.title('Розподіл суми рахунку відвідувачів ресторану')

# Створення підпису осі OX графіка
plt.xlabel('Сума рахунку, в доларах')

# Створення підпису осі OY графіка
plt.ylabel('Кількість рахунків')

# Виведення графіка
plt.show()
```

**2. Box plot чайових за днями тижня**

Створіть box plot, який показує розподіл чайових (`tip`) за днями тижня (`day`).
В який день навище медіанне (середнє) значення чайоивх? Зелена горизонтальна лінія всередині кожного box-plot.


```python
# Створення коробчатого графіка (box plot) розподілу чайових (tip) за днями тижня (day) для набору даних Tips
df.boxplot(column = 'tip', by = 'day')

# Створення назви графіка
plt.title('Розподіл чайових відвідувачів ресторану за днями тижня')

# Прибирання автоматичного заголовку графіка
plt.suptitle('')

# Створення підпису осі OX графіка
plt.xlabel('День тижня')

# Створення підпису осі OY графіка
plt.ylabel('Розмір чайових')

# Виведення графіка
plt.show()
```

```python
# В який день тижня найвище медіанне (середнє) значення чайових (зелена горизонтальна лінія всередині кожного box-plot)?
print("Медіанне (середнє) значення чайових найвище у неділю (Sun).")
```

**3. Scatter plot: рахунок vs чайові**

Створіть діаграму розсіювання для `total_bill` та `tip`.

```python
# Створення діаграми розсіювання (scatter plot) для колонок total_bill та tip набору даних Tips
df.plot.scatter(x = 'total_bill', y = 'tip')

# Створення назви графіка
plt.title('Залежність між розмірами рахунку та чайових відвідувачів ресторану')

# Створення підпису осі OX графіка
plt.xlabel('Сума рахунку')

# Створення підпису осі OY графіка
plt.ylabel('Розмір чайових')

# Виведення графіка
plt.show()
```

---

**4. Bar chart: середній рахунок за часом**

Створіть стовпчасту діаграму середнього рахунку за часом дня (`time`).

```python
# Створення стовпчастої діаграми (bar chart) середнього рахунку відвідувачів ресторану за часом дня (time) для набору даних Tips
df.groupby('time')['total_bill'].mean().plot(kind = 'bar')

# Створення назви графіка
plt.title('Середній рахунок відвідувачів ресторану за часом дня')

# Створення підпису осі OX графіка
plt.xlabel('Час дня')

# Створення підпису осі OY графіка
plt.ylabel('Середня сума рахунку')

# Виведення графіка
plt.show()
```

**5. Інтерактивна діаграма розсіювання**

Створіть з допомогою бібліотек Plotly інтерактивний scatter plot для `total_bill` vs `tip` з кольором за статтю (`sex`).


```python
# Створення за допомогою бібліотек Plotly інтерактивної діаграми розсіювання (scatter plot)
# для колонок total_bill та tip набору даних Tips з кольором за статтю (sex)

# Імпорт бібліотек Plotly під скороченими псевдонімами
import plotly.express as px
import plotly.graph_objects as go

# Створення інтерактивної діаграми розсіювання для колонок total_bill та tip з кольором за статтю (sex)
fig = px.scatter(df, x = 'total_bill', y = 'tip', color = 'sex', title = 'Залежність між сумою рахунку та чайовими за статтю',
                labels = {'total_bill': 'Сума рахунку', 'tip': 'Чайові', 'sex': 'Стать'})

# Виведення графіка
fig.show()
```

**6. Інтерактивний box plot: чайові за днями та статтю**

Створіть інтерактивний box plot для чайових за днями тижня з розбивкою за кольором по статі.

```python
# Створення інтерактивного коробчатого графіка (box plot) для чайових (tip) за днями тижня (day) з кольором за статтю (sex)
fig = px.box(df, x = 'day', y = 'tip', color = 'sex', title = 'Розподіл чайових за днями тижня та статтю',
                labels = {'day': 'День тижня', 'tip': 'Чайові', 'sex': 'Стать'})

# Виведення графіка
fig.show()
```
