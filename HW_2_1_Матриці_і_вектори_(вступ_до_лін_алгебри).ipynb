```python
# 1. Створення numpy вектора vector розмірністю 10 з випадковими дійсними значеннями в діапазоні від 0 до 1 двома способами

# Імпортування необхідних бібліотек
import numpy as np

# 1.1. Створення numpy вектора vector розмірністю 10 з випадковими дійсними значеннями в діапазоні від 0 до 1 способом генерації випадкових чисел
# з модулем random.uniform
vector = np.random.uniform(0, 1, 10)

# Виведення створеного вектора vector
print(f"vector = {vector}")

# Виведення розмірів вектора vector
print(f"Розміри вектора vector: {vector.shape}")
```


### Завдання 3: Множення матриць
1. Задані дві матриці `A` і `B`розмірностями 3x2 та 2х3. Виконайте їхнє множення, щоб отримати матрицю 2x2. Виведіть результуючу матрицю `result_matrix` на екран.

2. Заданий вектор `C` розмірністю 2х1. Подумайте спочатку, яку розмірність Ви отримаєте, коли домножите `result_matrix` на вектор `C` і який результат буде. А потім виконайте множення і виведіть результат на екран.

3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отримаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0, 1], [1, 0]])

display(A, B, C, D)
```

```python
# 3.1. Множення матриць A і B, щоб отримати результуючу матрицю result_matrix розмірності 2x2
result_matrix = np.dot(B, A)

# Виведення результату
print(f"Результуюча матриця result_matrix:")
print(result_matrix)
```

```python
# 3.2. Множення матриці result_matrix на вектор C
result_matrix_x_C = np.dot(result_matrix, C)

# Виведення результату
print(f"Результат множення матриці result_matrix на вектор C:")
print(result_matrix_x_C)
```

```python
# 3.3. Множення матриці result_matrix на матрицю D
result_matrix_x_D = np.dot(result_matrix, D)

# Виведення результату
print(f"Результат множення матриці result_matrix на матрицю D:")
print(result_matrix_x_D)
```

```python
# 4.3. Зміна форми вектора vector
column_vector = np.expand_dims(vector, axis = 1)

# Виведення результату
print(f"Вектор-стовпець vector:")
print(column_vector)

# Виведення форми вектора-стовпця vector
print(f"Розміри вектора-стовпця vector: {column_vector.shape}")

# Порівняння результату з vector
print("Форма вектора vector змінилася з рядка на стовпець, і тепер він має 2 виміри - 10 рядків та 1 стовпець.")
```
