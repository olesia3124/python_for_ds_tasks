**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:

```python
# 4.1. Обчислення стандартного відхилення для набору даних sum_purchases методами Python

# Обчислення різниць між елементами sum_purchases та його вибірковим середнім, піднесених до квадрату
sum_purchases_minus_mean_xx2 = [(el - mean_sum_purchases)**2 for el in sum_purchases]

# Обчислення стандартного відхилення вибірки sum_purchases
std_sum_purchases_sample = (sum(sum_purchases_minus_mean_xx2) / (len(sum_purchases) - 1)) ** 0.5
print(f"Стандартне відхилення для набору даних sum_purchases методами Python: {std_sum_purchases_sample}")
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
# 6.1. Обчислення коефіцієнта кореляції Пірсона для суми покупок (sum_purchases) і кількості відвідувань (visits) методами Python

# Знаходження вибіркового середнього для набору даних visits
mean_visits = sum(visits) / len(visits)
print(f"Вибіркове середнє для набору даних visits методами Python: {mean_visits}")

# Обчислення стандартного відхилення для набору даних visits

# Обчислення різниць між елементами visits та його вибірковим середнім
visits_minus_mean = [(el - mean_visits) for el in visits]

# Обчислення різниць між елементами visits та його вибірковим середнім, піднесених до квадрату
visits_minus_mean_xx2 = [(el - mean_visits)**2 for el in visits]

# Обчислення стандартного відхилення вибірки visits
std_visits_sample = (sum(visits_minus_mean_xx2) / (len(visits) - 1)) ** 0.5
print(f"Стандартне відхилення для набору даних visits методами Python: {std_visits_sample}")

# Обчислення різниць між елементами sum_purchases та його вибірковим середнім
sum_purchases_minus_mean = [(el - mean_sum_purchases) for el in sum_purchases]

# Обчислення коваріації для суми покупок (sum_purchases) і кількості відвідувань (visits)
cov_sum_purchases_visits = sum(x * y for x, y in zip(sum_purchases_minus_mean, visits_minus_mean)) / (len(sum_purchases) - 1)
print(f"Коваріація для суми покупок (sum_purchases) і кількості відвідувань (visits) методами Python: {cov_sum_purchases_visits}")

# Обчислення коефіцієнта кореляції Пірсона для суми покупок (sum_purchases) і кількості відвідувань (visits)
corr_Pierson_sum_purchases_visits = cov_sum_purchases_visits / (std_sum_purchases_sample * std_visits_sample)
print(f"Коефіцієнт кореляції Пірсона для суми покупок (sum_purchases) і кількості відвідувань (visits) методами Python: {corr_Pierson_sum_purchases_visits}")
```

```python
# 6.2. Обчислення коефіцієнта кореляції Пірсона для суми покупок (sum_purchases) і кількості відвідувань (visits) методами numpy
corr_Pierson_sum_purchases_visits_np = np.corrcoef(sum_purchases, visits)[0, 1]
print(f"Коефіцієнт кореляції Пірсона для суми покупок (sum_purchases) і кількості відвідувань (visits) методами numpy: {corr_Pierson_sum_purchases_visits_np}")
```

```python
# Порівняння знайдених коефіцієнтів кореляції Пірсона для суми покупок (sum_purchases) і кількості відвідувань (visits)
if corr_Pierson_sum_purchases_visits - corr_Pierson_sum_purchases_visits_np <= 1e-10:
    print("Значення коефіцієнта кореляції Пірсона співпадають!")
else:
    print("Значення коефіцієнта кореляції Пірсона не співпадають!")
```

Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
# Обчислення коефіцієнта кореляції Пірсона для суми покупок (sum_purchases) і кількості відвідувань (visits) після вилучення викидів

# Вилучення викидів з набору даних sum_purchases
threshold_sum_purchases = mean_sum_purchases_np + 2 * std_sum_purchases_sample_np
filtered_sum_purchases = [x for x in sum_purchases if x < threshold_sum_purchases]
print(f"Набір даних sum_purchases без викидів:{filtered_sum_purchases}")

# Вилучення викидів з набору даних visits
filtered_visits = [y for x, y in zip(sum_purchases, visits) if x < threshold_sum_purchases]
print(f"Набір даних visits без викидів:{filtered_visits}")

# Обчислення коефіцієнта кореляції Пірсона для sum_purchases і visits без викидів
corr_Pierson_filtered_sum_purchases_visits = np.corrcoef(filtered_sum_purchases, filtered_visits)[0, 1]
print(f"Коефіцієнт кореляції Пірсона для суми покупок (sum_purchases) і кількості відвідувань (visits) після вилучення викидів: {corr_Pierson_filtered_sum_purchases_visits}")
```
