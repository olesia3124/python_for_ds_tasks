**1. Функція з параметром**

Створіть функцію `show_menu()`, яка приймає перелік страв, наприклад ['Піца', 'Паста', 'Салат'] і виводить меню ресторану в наступному вигляді:
```
=== МЕНЮ ===
1. Піца
2. Паста
3. Салат
=============
```

Викличте цю функцію на списку ["Суп грибний", "Котлета по-київськи", "Лимонад"].

````python
# Функція для створення та виведення меню ресторану

# Означення функції, яка виводить заголовок меню, кожен елемент переліку страв разом з порядковим номером та кінець меню
def show_menu(dish_list):
    print("=== МЕНЮ ===")
    for i, dish in enumerate(dish_list):
        print(f"{i+1}. {dish}")
    print("=============")

# Список з переліком страв
dish_list = ["Суп грибний", "Котлета по-київськи", "Лимонад"]

# Виклик функції
show_menu(dish_list)
````

Наприклад: `print_square(5)` → "Квадрат числа 5 = 25"

````python
# Функція для виведення квадрату числа

# Означення функції, яка виводить квадрат числа
def print_square(n):
    print(f"Квадрат числа {n} = {n**2}")

# Виклик функції
print_square(5)
````

**4. Функція пошуку максимумального значення**

Створіть функцію `find_max(numbers)`, яка:
- Приймає список чисел
- Повертає найбільше число в списку

Протестуйте на списку `[15, 42, 8, 23, 56, 19]`.


````python
# Функція для знаходження максимального значення заданого списку

# Означення функції, яка знаходить максимальне значення заданого списку
def find_max(numbers):
    return max(numbers)

# Задання списку
numbers = [15, 42, 8, 23, 56, 19]

# Виклик функції
print(f"Максимальне значення: {find_max(numbers)}")
````

**5. Функція з умовами**

Створіть функцію `check_grade(grade)`, яка:
- Приймає оцінку (число)
- Повертає "Здав" якщо grade >= 60
- Повертає "Не здав" якщо grade < 60

Протестуйте на різних оцінках.


````python
# Функція для визначення задовільності оцінки

# Означення функції, яка приймає оцінку і повертає "Здав" при grade >= 60 або "Не здав" при grade < 60
def check_grade(grade):
    if grade >= 60:
        return("Здав")
    else:
        return("Не здав")

# Перевірка роботи функції для різних значень
grade = 81
print(f"Результат для оцінки {grade}: {check_grade(grade)}")

grade = 59
print(f"Результат для оцінки {grade}: {check_grade(grade)}")
````
