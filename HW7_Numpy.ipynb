**1. Створення масивів**

Створіть (кожен пункт оцінюється в 0.5 бала):

```python
# Імпорт бібліотеки numpy під скороченим псевдонімом np
import numpy as np

# Створення масиву із заданого списку
list = [5, 15, 25, 35, 45]
arr_from_list = np.array(list)
print(arr_from_list)
```

**2. Базові операції**

Створіть Numpy-масив зі значеннями [12, 24, 36, 48, 60].

Виконайте наступні операції над **оригінальним** масивом і запишіть у нову змінну результат (кожен пункт оцінюється в 0.5 бала):
- Помножте всі елементи на 2
- Додайте до всіх елементів 10
- Піднесіть всі елементи до квадрату
- Обчисліть квадратний корінь з усіх елементів

Виведіть результат виконання кожної операції.

```python
# Створення заданого масиву
arr = np.array([12, 24, 36, 48, 60])
print(f"Початковий масив: {arr}")

# Домноження всіх елементів масиву на 2
x2_arr = arr * 2
print(f"Масив, домножений на 2: {x2_arr}")

# Збільшення всіх елементів масиву на 10
arr_add_10 = arr + 10
print(f"Масив з елементами, збільшеними на 10: {arr_add_10}")

# Піднесення всіх елементів масиву до квадрату
xx2_arr = arr ** 2
print(f"Масив, піднесений до квадрату: {xx2_arr}")

# Обчислення квадратного кореня з усіх елементів масиву
sqrt_arr = np.sqrt(arr)
print(f"Масив після здобуття квадратного кореня з елементів: {sqrt_arr}")
```

**3. Статистичні функції**

Створіть Numpy-масив `data` зі значеннями [18, 27, 33, 41, 52, 68, 79].

Обчисліть використовуючи функціонал numpy (кожен пункт оцінюється в 0.5 бала):
- Середнє значення
- Мінімум
- Максимум
- Суму всіх елементів

```python
# Створення заданого масиву
data = np.array([18, 27, 33, 41, 52, 68, 79])
print(f"Масив: {data}")

# Обчислення середнього значення масиву
mean_data = np.mean(data)
print(f"Середнє значення масиву: {mean_data}")

# Обчислення мінімуму масиву
min_data = np.min(data)
print(f"Мінімум масиву: {min_data}")

# Обчислення максимуму масиву
max_data = np.max(data)
print(f"Максимум масиву: {max_data}")

# Обчислення суми всіх елементів масиву
sum_data = np.sum(data)
print(f"Сума всіх елементів масиву: {sum_data}")
```

**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3
- Елементи більші за 40

```python
# Створення заданого масиву
arr = np.array([15, 22, 35, 44, 58, 67])
print(f"Масив: {arr}")

# Виведення першого елемента масиву
print(f"Перший елемент масиву: {arr[0]}")

# Виведення останнього елемента масиву
print(f"Останній елемент масиву: {arr[-1]}")

# Виведення елементів масиву з індексами від 1 до 3
print(f"Елементи масиву з індексами від 1 до 3: {arr[1:4]}")

# Виведення елементів масиву, більших за 40
mask = arr > 40
print(f"Елементи масиву, більші за 40: {arr[mask]}")
```

**5. Двовимірні масиви**

Задано матрицю 3x3.

Виконайте наступні дії з функціоналом Numpy (кожен пункт оцінюється в 0.5 бала):
- Виведіть форму (shape) матриці
- Виведіть перший рядок
- Виведіть другу колонку
- Обчисліть суму всіх елементів
- Обчисліть середнє всіх елементів

```python
matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])
```

```python
# Виведення форми матриці
print(f"Форма матриці: {np.shape(matrix)}")

# Виведення першого рядка матриці
print(f"Перший рядок матриці: {matrix[0, :]}")

# Виведення другої колонки матриці
print(f"Друга колонка матриці: {matrix[:, 1]}")

# Обчислення суми всіх елементів матриці
print(f"Сума всіх елементів матриці: {np.sum(matrix)}")

# Обчислення середнього значення всіх елементів матриці
print(f"Середнє значення всіх елементів матриці: {np.mean(matrix)}")
```
