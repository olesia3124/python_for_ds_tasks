**1. Цикл for - базовий**

Використовуючи цикл `for`, виведіть числа від 1 до 7 включно використовуючи функцію range.


```python
# Програма для виведення чисел від 1 до 7
for i in range(7):
    print(i+1)
```


---

**2. Цикл for по списку**

Дано список:

```python
names = ["Олег", "Анна", "Іван", "Марія"]
```


Використовуючи цикл, виведіть кожне ім'я з привітанням: "Привіт, [ім'я]!"


```python
# Програма для виведення привітання для імен зі списку names
for name in names:
    print(f"Привіт, {name}!")
```

**3. Цикл з enumerate**

Дано список оцінок:

```python
grades = [85, 92, 78, 95, 88]
```


Використовуючи `enumerate`, виведіть кожну оцінку з номером: "Студент 1: 85 балів"


```python
# Програма для виведення оцінок студентів зі списку grades
for i, grade in enumerate(grades):
    print(f"Студент {i+1}: {grade} балів")
```

**4. Сума парних чисел**

Напишіть програму, яка обчислює суму всіх парних чисел від 1 до 20 включно.

```python
# Програма для обчислення суми всіх парних чисел від 1 до 20 включно

# Створення списку всіх парних чисел від 1 до 20 включно
even_numbers = [x+1 for x in range(20) if (x+1)%2 == 0]

# Виведення суми всіх парних чисел від 1 до 20 включно
print(f"Сума парних чисел: {sum(even_numbers)}")
```
