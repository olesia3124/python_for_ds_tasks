### Домашні завдання на відпрацювання поняття функції та її похідної

1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
# 1. Обчислення похідної функції f(x) = 10x - 5 з допомогою Python та побудова графіків f(x) й f'(x)

# Імпортування потрібних бібліотек
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# Визначення символічної змінної x
x_sym = sp.symbols('x')

# Задання функції f(x) = 10x - 5 для обчислення похідної
f_sym = 10 * x_sym - 5

# Знаходження похідної f'(x)
df_sym = sp.diff(f_sym, x_sym)

# Виведення обчисленої похідної f'(x)
print("Похідна f'(x) функції f(x) = 10x - 5:", df_sym)

# Генерація лінійного простору зі 100 точок від 0 до 10 для осі 0x
x = np.linspace(0, 10, 100)

# Задання функції f(x) = 10x - 5 на згенерованому лінійному просторі
f = 10 * x - 5

# Задання похідної f'(x) = 10 (масиву з десяток)
df = np.ones_like(x) * 10

# Побудова графіку функції f(x) = 10x - 5
plt.plot(x, f, label = "f(x) = 10x - 5", color = 'blue')

# Побудова графіку функції f'(x)
plt.plot(x, df, label = "f'(x) = 10", color = 'green')

# Встановлення діапазону осі 0x від 0 до 10
plt.xlim(0, 10)

# Створення заголовка, міток, легенди та сітки графіків
plt.title("Графік функції f(x) = 10x - 5 та її похідної f'(x)")
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

# Виведення графіків
plt.show()

# Аналіз впливу зміни кількості годин вивчення на оцінку
print("""Зі зростанням кількості вивчених годин зростає й оцінка, зокрема, збільшення кількості вивчених годин на 1 збільшує оцінку на 10.""")
```

```python
# 2.2. Обчислення похідної функції f(x) = ln(x) і побудова графіків функції та її похідної

# Задання функції f(x) = ln(x) для обчислення похідної
f_sym = sp.ln(x_sym)

# Знаходження похідної f'(x)
df_sym = sp.diff(f_sym, x_sym)

# Виведення обчисленої похідної f'(x)
print("Похідна f'(x) функції f(x) = ln(x):", df_sym)

# Генерація лінійного простору зі 100 точок від 0.1 до 10 для осі 0x, щоб уникнути ділення на 0
x = np.linspace(0.1, 10, 100)

# Задання функції f(x) = ln(x) на лінійному просторі, згенерованому попередньо
f = np.log(x)

# Задання похідної f'(x) = 1 / x
df = 1 / x

# Побудова графіку функції f(x) = ln(x)
plt.plot(x, f, label = "f(x) = ln(x)", color = 'blue')

# Побудова графіку функції f'(x)
plt.plot(x, df, label = "f'(x) = 1 / x", color = 'green')

# Встановлення діапазону осі 0x від 0 до 10
plt.xlim(0, 10)

# Створення заголовка, міток, легенди та сітки графіків
plt.title("Графік функції f(x) = ln(x) та її похідної f'(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

# Виведення графіків
plt.show()
```

```python
# 3. Обчислення похідної функції f(x) = x^2 - 5x + 1 і побудова графіків функції та її похідної з допомогою Python та SymPy

# Задання функції f(x) = x^2 - 5x + 1 для обчислення похідної
f_sym = x_sym ** 2 - 5 * x_sym + 1

# Знаходження похідної f'(x)
df_sym = sp.diff(f_sym, x_sym)

# Виведення обчисленої похідної f'(x)
print("Похідна f'(x) функції f(x) = x^2 - 5x + 1:", df_sym)

# Генерація лінійного простору зі 100 точок від 0 до 100 для осі 0x
x = np.linspace(0, 100, 1000)

# Задання функції f(x) = x^2 - 5x + 1 на згенерованому лінійному просторі
f = x ** 2 - 5 * x + 1

# Задання похідної f'(x) = 2x - 5
df = 2 * x - 5

# Побудова графіку функції f(x) = x^2 - 5x + 1
plt.plot(x, f, label = "f(x) = x^2 - 5x + 1", color = 'blue')

# Побудова графіку функції f'(x)
plt.plot(x, df, label = "f'(x) = 2x - 5", color = 'green')

# Встановлення діапазону осі 0x від 0 до 100
plt.xlim(0, 100)

# Створення заголовка, міток, легенди та сітки графіків
plt.title("Графік функції f(x) = x^2 - 5x + 1 та її похідної f'(x)")
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)

# Виведення графіків
plt.show()
```
