**1. Списки - базові операції**

Створіть список температур за тиждень (у градусах Цельсія) з назвою
`temperatures` і значеннями [22, 25, 19, 23, 26, 20, 21, 22].


Виконайте (кожен пункт оцінюється в 0.5 бала):
- Виведіть перший елемент
- Виведіть останній елемент
- Додайте температуру 24
- Видаліть температуру 19
- Виведіть довжину списку
- Обчисліть суму всіх температур
- Знайдіть максимальну температуру

```python
# Список температур за тиждень (у градусах Цельсія)
temp_over_week = [22, 25, 19, 23, 26, 20, 21, 22]

# Перший елемент списку
print("Перший елемент:", temp_over_week[0])

# Останній елемент списку
print("Останній елемент:", temp_over_week[-1])

# Додавання температури 24
temp_over_week.append(24)

# Видалення температури 19
temp_over_week.remove(19)

# Виведення довжини списку
print("Довжина списку:", len(temp_over_week))

# Обчислення суми всіх температур
print("Сума:", sum(temp_over_week))

# Знаходження максимальної температури
print("Максимум:", max(temp_over_week))
```


**2. Словники - базові операції**

Створіть словник про себе, який містить ключі `name`, `age`, `city` і значення, які відповідають вашему імені, віку і місту (можна вставити будь-які значення, не обовʼязково свої).


Виконайте (кожен пункт оцінюється в 0.5 бала):
- Виведіть значення ключа "name"
- Змініть значення ключа "age" на 21
- Додайте новий ключ "hobby" зі значенням "programming"
- Виведіть всі ключі
- Виведіть всі значення


```python
# Створення словника
dict = {
    "name" : "Олеся",
    "age" : 32,
    "city" : "Кам'янець-Подільський"
}

# Виведення значення ключа "name"
print(dict["name"])

# Зміна значення ключа "age" на 21
dict["age"] = 21

# Додавання нового ключа "hobby" зі значенням "programming"
dict["hobby"] = "programming"

# Виведення всіх ключів
print(dict.keys())

# Виведення всіх значень
print(dict.values())
```
