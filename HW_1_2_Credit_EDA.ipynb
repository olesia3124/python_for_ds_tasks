# Робота з пропущеними значеннями

Найпростіше емпіричне правило для опрацювання пропущених значень: якщо пропущених значень більше за 40% - видаляємо колонку, якщо менше за 40% - аналізуємо, як можна заповнити і чи треба.

Зазвичай, якщо відсоток пропущених даних більший за 10 і немає чіткої (яка значно виділяється) моди (найчастішого) значення в даних, то лишаємо дані як є до використання методів машинного навчання. Далі, якщо метод вимагає заповнення пропущених значень, можемо експериментувати із заповненнями. Також хорошою практикою є створити окрему колонку-флаг (0/1), яка вказує, де були пропущені значення - це буде додаткова ознака для моделі.

## Кількість та відсоток пропущених значень у стовпцях

```python
# Обчислення кількості пропущених значень у стовпцях набору даних df
null_count = df.isnull().sum()
# Обчислення відсотку пропущених значень у стовпцях набору даних df
null_percentage = round((df.isnull().sum()/df.shape[0])*100, 2)
```

```python
# Створення датафрейму null_df пропущених значень набору даних df
null_df = pd.DataFrame({'column_name' : df.columns,'null_count' : null_count,'null_percentage': null_percentage})
# Видалення індексу та його заміна на новий індекс
null_df.reset_index(drop = True, inplace = True)
```

```python
# Виведення відсортованого за відсотком пропущених значень в порядку спадання датафрейму null_df
null_df.sort_values(by = 'null_percentage', ascending = False)
```

## Видалення стовпців з NULL значеннями > 40%

Сформуємо список стовпців з NULL значеннями > 40% у список. Ми видалимо ці стовпці з датафрейму, оскільки в них занадто багато пропущених значень.

```python
# Створення списку стовпців з відсотком пропущених значень, більшим за 40%
columns_to_be_deleted = null_df[null_df['null_percentage'] > 40].column_name.to_list()
```

```python
# Підсумовування кількості стовпців з відсотком пропущених значень, більшим за 40%, шляхом визначення довжини списку назв таких стовпців
len(columns_to_be_deleted)
```

Всього потрібно видалити `49` стовпців. Видалення їх з основного датафрейму **`df`**

```python
# Видалення кількості стовпців з відсотком пропущених значень, більшим за 40%, з оновленням набору даних df
df.drop(columns = columns_to_be_deleted, inplace = True)
```

Перевірка підрахунку стовпців після видалення. Мало б залишитись лише `73` стовпці

# ДЗ 2. Аналіз і заповнення пустих значень у колонках з малим відсотком пропущених значень

За прикладом вище проведіть аналіз пропущених значень в колонках
- EXT_SOURCE_2
- AMT_GOODS_PRICE

Для швидшого аналізу рекомендую написати фукнцію, яку Ви зможете викликати для кожної з колонок.

Зробіть висновок, що робити з пропущеними значеннями в кожному випадку, і виконайте ту дію, яку зазначили.

```python
# Функція для аналізу пропущених значень у стовпцях з малим відсотком таких значень
def analyze_column_null_df_under1(column_name, df, null_df_under1):
    """Аналіз та візуалізація стовпця з відсотком пропущених значень від 0% до 1%"""
    print(f"=== АНАЛІЗ СТОВПЦЯ {column_name} ===")

    # Відсоток пропущених значень з null_df_under1
    print("\n--- Інформація про пропущені значення ---")
    null_info = null_df_under1[null_df_under1.column_name == column_name]
    if null_info.empty:
        print(f"Записів для '{column_name}' у null_df_under1 не знайдено.")
    else:
        print(null_info)

    # Загальна статистика для будь-яких типів значень (мода)
    print("\n--- Базова статистика ---")
    print(f"Мода: {df[column_name].mode()[0]}")

    # Розгалуження аналізу залежно від типу даних
    if pd.api.types.is_numeric_dtype(df[column_name]):
        print("\nТип стовпця: Числовий")

        # Середнє та медіана
        print(f"Середнє значення: {df[column_name].mean()}")
        print(f"Медіана: {df[column_name].median()}")

        # Процентилі
        print("\nКвартилі (25%, 50%, 75%, 100%):")
        print(df[column_name].quantile(q = [0.25, 0.5, 0.75, 1]))

        # Візуалізація: Коробчатий графік (box plot)
        # Задання розмірів коробчатого графіка
        plt.figure(figsize = (10, 4))
        # Побудова горизонтального коробчатого графіка для кращого огляду на екрані
        df[column_name].plot(kind = "box", vert = False)
        plt.title(f"Boxplot для {column_name}")
        # Додання сітки по осі x для зручного огляду проекцій її значень на box plot
        plt.grid(axis = "x", linestyle = "--", alpha = 0.7)
        # Виведення коробчатого графіка
        plt.show()

    else:
        print("\nТип стовпця: Категоріальний")

        # Підрахунок кількості значень
        counts = df[column_name].value_counts()
        print("\nКількість зустрічань кожного значення:")
        print(counts)

        # Візуалізація: Стовпчаста діаграма (bar chart)
        # Задання розмірів стовпчастої діаграми
        plt.figure(figsize = (10, 5))
        # Побудова стовпчастої діаграми
        counts.plot(kind = "bar")
        # Повернення підписів значень під кутом 90 градусів для кращої читабельності
        plt.xticks(rotation = 90)
        # Заголовок стовпчастої діаграми
        plt.title(f"Розподіл значень для {column_name}")
        # Підпис осі OY
        plt.ylabel("Кількість зустрічань кожного значення")
        # Виведення стовпчастої діаграми
        plt.show()
```

```python
# Виклик функції аналізу пропущених значень для стовпця EXT_SOURCE_2
analyze_column_null_df_under1('EXT_SOURCE_2', df, null_df_under1)
```

### Створення функції `age_cat` для категоризації `YEARS_BORNING`

```python
def age_cat(years):
    if years <= 20:
        return '0-20'
    elif years > 20 and years <= 30:
        return '20-30'
    elif years > 30 and years <= 40:
        return '30-40'
    elif years > 40 and years <= 50:
        return '40-50'
    elif years > 50 and years <= 60:
        return '50-60'
    elif years > 60 and years <= 70:
        return '60-70'
    elif years > 70:
        return '70+'
```

### Одномірний (категоріальний vs цільовий) та двомірний (категоріальний vs категоріальний) аналіз (гістограма) - категоріальні стовпчики

```python
import matplotlib.pyplot as plt


def bi_cat_countplot(df, column, hue_column):
    unique_hue_values = df[hue_column].unique()
    fig, axes = plt.subplots(nrows=1, ncols=2)
    fig.set_size_inches(14,6)

    pltname = f'Нормалізований розподіл значень за категорією: {column}'
    proportions = df.groupby(hue_column)[column].value_counts(normalize=True)
    proportions = (proportions*100).round(2)
    ax = proportions.unstack(hue_column).sort_values(
        by=unique_hue_values[0], ascending=False
        ).plot.bar(ax=axes[0], title=pltname)

    # анотація значень в барплоті
    for container in ax.containers:
        ax.bar_label(container, fmt='{:,.1f}%')

    pltname = f'Кількість даних за категорією: {column}'
    counts = df.groupby(hue_column)[column].value_counts()
    ax = counts.unstack(hue_column).sort_values(
        by=unique_hue_values[0], ascending=False
        ).plot.bar(ax=axes[1], title=pltname)

    for container in ax.containers:
      ax.bar_label(container)


def uni_cat_target_compare(df, column):
    bi_cat_countplot(df, column, hue_column='TARGET' )


def bi_countplot_target(df0, df1, column, hue_column):
  pltname = 'Клієнт зі складнощами щодо платності'
  print(pltname.upper())
  bi_cat_countplot(df1, column, hue_column)
  plt.show()

  pltname = 'Клієнти зі своєчасними платежами'
  print(pltname.upper())
  bi_cat_countplot(df0, column, hue_column)
  plt.show()
```

### Обчислення мінімального та максимального діапазону відхилень для числових стовпців

```python
def outlier_range(dataset,column):
    Q1 = dataset[column].quantile(0.25)
    Q3 = dataset[column].quantile(0.75)
    IQR = Q3 - Q1
    Min_value = (Q1 - 1.5 * IQR)
    Max_value = (Q3 + 1.5 * IQR)
    return Max_value
```

# Робота з викидами для числових стовпців

```python
import warnings

def dist_box(dataset, column):
    with warnings.catch_warnings():
      warnings.simplefilter("ignore")

      plt.figure(figsize=(16,6))

      plt.subplot(1,2,1)
      sns.distplot(dataset[column], color = 'purple')
      pltname = 'Графік розподілу для ' + column
      plt.ticklabel_format(style='plain', axis='x')
      plt.title(pltname)

      plt.subplot(1,2,2)
      red_diamond = dict(markerfacecolor='r', marker='D')
      sns.boxplot(y = column, data = dataset, flierprops = red_diamond)
      pltname = 'Боксплот для ' + column
      plt.title(pltname)

      plt.show()
```

```python
# Створення функції analyze_column_outliers для проведення одновимірного аналізу стовпця двох масивів даних
def analyze_column_outliers(df0, df1, column):
    """Функція для розрахунку викидів через IQR та побудови графіка KDE для стовпця двох масивів даних.
    """
    print(f"=== АНАЛІЗ СТОВПЦЯ: {column} ===")

    # Пошук викидів для df1 (Труднощі з оплатою)
    q1_1 = df1[column].quantile(0.25)
    q3_1 = df1[column].quantile(0.75)
    iqr_1 = q3_1 - q1_1
    min_value1 = q1_1 - 1.5 * iqr_1
    max_value1 = q3_1 + 1.5 * iqr_1

    print("\n[Payment difficulties]")
    print(f"IQR: {iqr_1}")
    print(f"Мінімальне значення без викидів: {min_value1}")
    print(f"Максимальне значення без викидів: {max_value1}")

    # Пошук викидів для df0 (Вчасні оплати)
    q1_0 = df0[column].quantile(0.25)
    q3_0 = df0[column].quantile(0.75)
    iqr_0 = q3_0 - q1_0
    min_value0 = q1_0 - 1.5 * iqr_0
    max_value0 = q3_0 + 1.5 * iqr_0

    print("\n[On-Time Payments]")
    print(f"IQR: {iqr_0}")
    print(f"Мінімальне значення без викидів: {min_value0}")
    print(f"Максимальне значення без викидів: {max_value0}")

    # Візуалізація даних без викидів (функція kde_no_outliers)
    
    # Налаштування розміру графіка (14 дюймів завширшки, 6 заввишки)
    plt.figure(figsize = (14, 6))

    # Фільтрація обох датафреймів за нижньою та верхньою межами викидів
    df1_filtered = df1[(df1[column] >= min_value1) & (df1[column] <= max_value1)]
    df0_filtered = df0[(df0[column] >= min_value0) & (df0[column] <= max_value0)]
    
    # Побудова кривої розподілу для клієнтів із труднощами оплати
    sns.kdeplot(df1_filtered[column], label="Payment difficulties")
    # Побудова кривої розподілу для клієнтів зі вчасною оплатою
    sns.kdeplot(df0_filtered[column], label="On-Time Payments")
    # Відключення наукового формату чисел для осі X, щоб показувати звичайні числа
    plt.ticklabel_format(style = "plain", axis = "x")
    # Повернення підписів осі X на 45 градусів для кращої читабельності
    plt.xticks(rotation = 45)
    # Виведення назви графіку
    plt.title(f"Розподіл {column} без викидів (IQR метод)")
    # Виведення легенди
    plt.legend()
    # Виведення графіку
    plt.show()
```

```python
# Одновимірний аналіз стовпця YEARS_BIRTH через виклик функції analyze_column_outliers
analyze_column_outliers(df0, df1, 'YEARS_BIRTH')
```

```python
# Створення функції кореляційного аналізу двох стовпців без викидів

def compare_two_columns(df0, df1, col_x, col_y, outlier_range):
    """Функція для здійснення кореляційного аналізу двох числових стовпців (scatter plot) для двох груп клієнтів із попереднім видаленням викидів.
    """
    # Розрахунок меж викидів за допомогою функції outlier_range
    # Пошук верхньої межі викидів для стовпця col_x у групі з платіжними труднощами (df1)
    max_x1 = outlier_range(df1, col_x)
    # Пошук верхньої межі викидів для стовпця col_y у групі з платіжними труднощами (df1)
    max_y1 = outlier_range(df1, col_y)
    # Пошук верхньої межі викидів для стовпця col_x у групі з вчасними оплатами (df0)
    max_x0 = outlier_range(df0, col_x)
    # Пошук верхньої межі викидів для стовпця col_y у групі з вчасними оплатами (df0)
    max_y0 = outlier_range(df0, col_y)

    # Друк розрахованих меж викидів для контролю
    print(f"=== МЕЖІ ВИКИДІВ ДЛЯ {col_x} VS {col_y} ===")
    print(f"[Payment difficulties] Max {col_x}: {max_x1:.2f} | Max {col_y}: {max_y1:.2f}")
    print(f"[On-Time Payments]     Max {col_x}: {max_x0:.2f} | Max {col_y}: {max_y0:.2f}\n")

    # Одночасна фільтрація обох ознак для уникнення розсинхронізації розміру масивів
    df1_filtered = df1[(df1[col_x] < max_x1) & (df1[col_y] < max_y1)]
    df0_filtered = df0[(df0[col_x] < max_x0) & (df0[col_y] < max_y0)]

    # Побудова графіків розсіювання (scatter plot)

    # Налаштування розміру вікна для графіків (14 дюймів завширшки, 6 заввишки)
    plt.figure(figsize = (14, 6))

    # Графік 1: Клієнти з труднощами оплати

    # Створення першого графіка у сітці (1 рядок, 2 стовпці, позиція 1)
    plt.subplot(1, 2, 1)
    # Створення заголовку для першої діаграми розсіювання
    plt.title("Payment difficulties")
    # Побудова діаграми розсіювання для клієнтів із труднощами оплати за відфільтрованими даними
    sns.scatterplot(x = col_x, y = col_y, data = df1_filtered, alpha = 0.6)
    # Відключення наукового формату чисел для осі X
    plt.ticklabel_format(style = "plain", axis = "x")
    # Відключення наукового формату чисел для осі Y
    plt.ticklabel_format(style = "plain", axis = "y")

    # Графік 2: Клієнти зі вчасною оплатою

    # Створення другого графіка у сітці (1 рядок, 2 стовпці, позиція 2)
    plt.subplot(1, 2, 2)
    # Створення заголовку для другої діаграми розсіювання
    plt.title("On-Time Payments")
    # Побудова діаграми розсіювання клієнтів зі вчасною оплатою за відфільтрованими даними
    sns.scatterplot(x = col_x, y = col_y, data = df0_filtered, alpha = 0.6)
    # Відключення наукового формату чисел для осі X
    plt.ticklabel_format(style = "plain", axis = "x")
    # Відключення наукового формату чисел для осі Y
    plt.ticklabel_format(style = "plain", axis = "y")

    # Налаштування відступів між двома графіками, щоб підписи не накладалися
    plt.tight_layout(pad = 4)
    # Виведення графіків
    plt.show()
```

## Неперервні та категоріальні змінні

```python
def draw_boxplot(df, categorical, continuous, max_continuous, title, hue_column, subplot_position):
    """
    Малює блок-діаграму для заданого DataFrame, категоріальної та неперервної змінної.
    """
    # Створення підграфіка у сітці (1 рядок, 2 стовпці, позиція визначається змінною subplot_position)
    plt.subplot(1, 2, subplot_position)
    # Встановлення заголовка для поточного підграфіка
    plt.title(title)
    # Налаштування стилю відображення викидів: червоні (r) ромби (D)
    red_diamond = dict(markerfacecolor = 'r', marker = 'D')
    # Побудова коробчастого графіка із фільтрацією викидів у числових даних та групуванням за категоріями
    sns.boxplot(x = categorical, # Категоріальна змінна для осі X
                y = df[df[continuous] < max_continuous][continuous], # Безперервна змінна для осі Y (відфільтрована за max_continuous)
                data = df, # Джерело даних (датафрейм)
                flierprops = red_diamond, # Застосування налаштованого стилю для точок-викидів
                order = sorted(df[categorical].unique(), reverse = True), # Сортування категорій на осі X у зворотному алфавітному порядку
                hue = hue_column, hue_order = sorted(df[hue_column].unique(), reverse = True)) # Додаткове групування
    # та розділення кольором за третьою колонкою
    # Відключення наукового формату чисел для осі Y
    plt.ticklabel_format(style = 'plain', axis = 'y')
    # Повернення підписів осі X на 90 градусів для кращої читабельності
    plt.xticks(rotation = 90)

def bi_boxplot(categorical, continuous, max_continuous1, max_continuous0, hue_column):
    """
    Створює паралельні блок-діаграми для двох груп, визначених у наборі даних, на основі
    категоріальної та неперервної змінної, виділяючи відмінності за допомогою відтінків.
    """
    # Налаштування розміру вікна для графіків (16 дюймів завширшки, 10 заввишки)
    plt.figure(figsize = (16, 10))

    # Графік для першої групи "Труднощі з платежами" (Payment Difficulties)
    draw_boxplot(df1, categorical, continuous, max_continuous1, 'Payment Difficulties', hue_column, 1)

    # Графік для другої групи "Вчасні оплати" (On-Time Payments)
    draw_boxplot(df0, categorical, continuous, max_continuous0, 'On-Time Payments', hue_column, 2)

    # Налаштування відступів між двома графіками, щоб підписи не накладалися
    plt.tight_layout(pad = 4)
    # Виведення графіків
    plt.show()
```

## Категоріальні та категоріальні змінні

### Аналіз стовпців `NAME_INCOME_TYPE` vs `CODE_GENDER`

```python
def bi_countplot_target(df0, df1, column, hue_column):
    # ПОБУДОВА НОРМАЛІЗОВАНИХ ГРАФІКІВ (ВІДСОТКИ)
    group_name = f"Нормалізований розподіл значень за категорією: {column}"
    print(group_name.upper())

    pltname = 'Клієнти зі складнощами щодо оплати'
    unique_hue_values = df1[hue_column].unique()
    # Ініціалізація вікна для першої пари графіків (відсотки)
    fig, axes = plt.subplots(nrows = 1, ncols = 2)
    # Налаштування розміру вікна для графіків (14 дюймів завширшки, 4 заввишки)
    fig.set_size_inches(14, 4)

    # Обчислення відсоткових пропорцій для групи з труднощами щодо оплати (df1)
    proportions = df1.groupby(hue_column)[column].value_counts(normalize = True)
    proportions = (proportions * 100).round(2)
    ax = proportions.unstack(hue_column).sort_values(
      by = unique_hue_values[0], ascending = False
      ).plot.bar(ax = axes[0], title = pltname)

    # Додавання текстових міток із відсотками поверх стовпчиків першого графіка
    for container in ax.containers:
      ax.bar_label(container, fmt = '{:,.1f}%')

    pltname = 'Клієнти зі своєчасними платежами'
    unique_hue_values = df0[hue_column].unique()

    # Обчислення відсоткових пропорцій для групи зі вчасною оплатою (df0)
    proportions = df0.groupby(hue_column)[column].value_counts (normalize = True)
    proportions = (proportions * 100).round(2)
    ax = proportions.unstack(hue_column).sort_values(
    by = unique_hue_values[0], ascending = False
      ).plot.bar(ax = axes[1], title = pltname)

    # Додавання текстових міток із відсотками поверх стовпчиків другого графіка
    for container in ax.containers:
      ax.bar_label(container, fmt = '{:,.1f}%')

    # Виведення першого блоку графіків
    plt.show()

    # ПОБУДОВА АБСОЛЮТНИХ ГРАФІКІВ (КІЛЬКІСТЬ КЛІЄНТІВ) 
    group_name = f"Кількість значень за категорією {column}"
    print(group_name.upper())

    pltname = 'Клієнти зі складнощами щодо оплати'
    unique_hue_values = df1[hue_column].unique()
    # Ініціалізація вікна для другої пари графіків (абсолютна кількість)
    fig, axes = plt.subplots(nrows = 1, ncols = 2)
    # Налаштування розміру вікна для графіків (14 дюймів завширшки, 4 заввишки)
    fig.set_size_inches(14, 4)

    # Підрахунок кількості клієнтів у кожній підгрупі з труднощами щодо оплати (df1)
    counts = df1.groupby(hue_column)[column].value_counts()
    ax = counts.unstack(hue_column).sort_values(
      by = unique_hue_values [0], ascending = False
      ).plot.bar(ax = axes [0], title = pltname)

    # Додавання міток із точною кількістю клієнтів поверх стовпчиків третього графіка
    for container in ax.containers:
      ax.bar_label(container)

    pltname = 'Клієнти зі своєчасними платежами'
    unique_hue_values = df0[hue_column].unique()
    # Підрахунок кількості клієнтів у кожній підгрупі зі вчасною оплатою (df0)
    counts = df0.groupby(hue_column)[column].value_counts()
    ax = counts.unstack(hue_column).sort_values(
      by = unique_hue_values[0], ascending = False
      ).plot.bar(ax = axes[1], title = pltname)

    # Додавання міток із точною кількістю клієнтів поверх стовпчиків четвертого графіка
    for container in ax.containers:
      ax.bar_label(container)

    # Виведення другого блоку графіків
    plt.show()
```

```python
# Аналіз взаємозалежностей між двома категоріальними змінними NAME_INCOME_TYPE та CODE_GENDER через виклик функції
# bi_countplot_target
bi_countplot_target(df0, df1, 'NAME_INCOME_TYPE', 'CODE_GENDER')
```

**Спостереження**
